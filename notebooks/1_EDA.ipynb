# 1. EDA — Olist Brazilian E-Commerce

**Dataset:** ~100k orders across 8 relational tables (2016–2018).

```python
from __future__ import annotations

import sys
from pathlib import Path

import pandas as pd

sys.path.insert(0, str(Path("..").resolve()))
from src.paths import RAW_FILES, raw_path
```

## Load all datasets

```python
dfs: dict[str, pd.DataFrame] = {}

for name in RAW_FILES:
    dfs[name] = pd.read_csv(raw_path(name))
    print(f"{name:<22} {dfs[name].shape}")
```

## Schema overview

```python
for name, df in dfs.items():
    print(f"\n{'─' * 55}")
    print(f"  {name.upper()}  ({df.shape[0]:,} rows × {df.shape[1]} cols)")
    print(f"{'─' * 55}")
    display(df.dtypes.rename("dtype").to_frame())
```

## Missing values

```python
for name, df in dfs.items():
    missing = df.isna().sum()
    missing = missing[missing > 0]
    if not missing.empty:
        print(f"\n{name}")
        display(missing.to_frame("missing").assign(pct=lambda x: (x["missing"] / len(df) * 100).round(1)))
```

## Sample rows

```python
for name, df in dfs.items():
    print(f"\n{name}")
    display(df.head(3))
```

## Orders — time range & status breakdown

```python
orders = dfs["orders"].copy()
orders["order_purchase_timestamp"] = pd.to_datetime(orders["order_purchase_timestamp"])

print("Date range:", orders["order_purchase_timestamp"].min(), "→", orders["order_purchase_timestamp"].max())
display(orders["order_status"].value_counts().to_frame())
```
